```python
dep = cv2.imread("dept.jpg")
```

```python
print(dep.dtype)
```

```python
dep = tf.cast(dep,tf.float32)/255
dep = tf.expand_dims(dep,axis=0)
```

```python
print(dep.shape)
```

```python
import numpy as np
```

```python
kernel=np.array([[0,-1,0],[-1,4,-1],[0,-1,0]], dtype = np.float32)
```

```python
print(kernel)
```

```python
kernel = kernel[:,:,np.newaxis,np.newaxis]
```

```python
kernel = np.repeat(kernel,repeats=3,axis=2)
```

```python
kernel = np.repeat(kernel,repeats=3,axis=3)
```

```python
print(kernel.shape)
```

```python
param = Conv2D(filters=3,strides=(1,1), padding="same", kernel_size=(3,3),use_bias=False)
```

```python
param.build(input_shape=dep.shape)
```

```python
param.set_weights([kernel])
```

```python
output = param(dep)[0]
```

```python
plt.imshow(output)
```
